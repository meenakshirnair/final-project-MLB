```python
# Cell 1: Import Libraries and Load Data
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

# Load the dataset
df = pd.read_csv('/attrition_balanced.csv')
# Label encode all object-type columns
df = df.apply(lambda col: LabelEncoder().fit_transform(col) if col.dtypes == 'object' else col)

# Drop redundant ID-like columns
df.drop(['EmployeeCount', 'EmployeeNumber', 'StandardHours', 'Over18'], axis=1, inplace=True)

# Features and target
X = df.drop('Attrition', axis=1)
y = df['Attrition']

# Min-Max Scaling
scaler = MinMaxScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, stratify=y, random_state=42)

df.head()
```

```python
# Cell 2: Define Evaluation Function
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score

def evaluate_model(name, model, X_test, y_test):
    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        from sklearn.calibration import CalibratedClassifierCV
        model = CalibratedClassifierCV(model, cv="prefit")
        model.fit(X_train, y_train)
        y_prob = model.predict_proba(X_test)[:, 1]
    return {
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "AUC": roc_auc_score(y_test, y_prob)
    }
```

```python
# Cell 3: Train SVM with Hyperparameter Tuning
from sklearn.svm import SVC
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

options = [
    {'C': 0.1, 'kernel': 'linear'},
    {'C': 1, 'kernel': 'rbf'},
    {'C': 10, 'kernel': 'rbf'},
    {'C': 100, 'kernel': 'rbf'}
]

for params in options:
    model = SVC(**params, gamma='scale', probability=True)
    model.fit(X_train, y_train)
    print(f"\nParams: {params}")
    print(evaluate_model(f"SVM {params}", model, X_test, y_test))

# Confusion Matrix
ConfusionMatrixDisplay.from_estimator(model, X_test, y_test, cmap='Blues')
plt.title("Confusion Matrix: SVM")
plt.show()
```

```python
# Cell 3 Backup: Train SVM with Hyperparameter Tuning using GridSearchCV
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import classification_report, ConfusionMatrixDisplay, roc_curve, auc
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
import numpy as np

# Define SVM hyperparameter grid
svm_params = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf']
}

# Set up GridSearchCV
svm = SVC(gamma='scale', probability=True, random_state=42)
svm_grid = GridSearchCV(svm, svm_params, cv=5, scoring='accuracy')
svm_grid.fit(X_train, y_train)

# Get the best model
best_svm = svm_grid.best_estimator_
print("✅ SVM Best Params:", svm_grid.best_params_)
svm_results = evaluate_model("SVM", best_svm, X_test, y_test)
print("SVM Tree Best Params:", svm_grid.best_params_)
print(svm_results)

# Evaluate the best SVM
y_pred = best_svm.predict(X_test)
print(classification_report(y_test, y_pred))

# Confusion Matrix
ConfusionMatrixDisplay.from_estimator(best_svm, X_test, y_test, cmap='Blues')
plt.title("Confusion Matrix: Best SVM Model")
plt.show()

# ROC Curve
y_proba = best_svm.predict_proba(X_test)[:, 1]
fpr, tpr, _ = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f'Best SVM (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve: Best SVM Model")
plt.legend(loc="lower right")
plt.grid()
plt.show()

# Feature Importance using Permutation Importance
result = permutation_importance(best_svm, X_test, y_test, n_repeats=10, random_state=42)

# Sort and plot top N features
top_n = 10
sorted_idx = result.importances_mean.argsort()[::-1]
plt.figure(figsize=(10, 6))
plt.barh(range(top_n), result.importances_mean[sorted_idx][:top_n], align='center')
plt.yticks(range(top_n), [X_train.columns[i] for i in sorted_idx[:top_n]])
plt.gca().invert_yaxis()
plt.title("Top 10 Feature Importances (Permutation Importance)")
plt.xlabel("Mean Importance Score")
plt.show()
```

```python
# Import necessary libraries
from sklearn.linear_model import LogisticRegression  # Import the logistic regression algorithm
from sklearn.feature_selection import SequentialFeatureSelector  # Import tool for sequential feature selection
from sklearn.pipeline import make_pipeline  # Import function to create processing pipelines
from sklearn.preprocessing import StandardScaler  # Import scaling tool for data normalization

# Initialize logistic regression model with increased maximum iterations
# This helps ensure the model converges to a solution
logreg = LogisticRegression(max_iter=1000)

# Forward Feature Selection
# -------------------------
# Initialize forward selection strategy
# This will start with no features and add one feature at a time based on performance
# cv=5 indicates 5-fold cross-validation to evaluate feature contributions
forward_selector = SequentialFeatureSelector(logreg, direction='forward', cv=5)

# Train the feature selector on training data
# This process identifies which features to keep based on their predictive power
forward_selector.fit(X_train, y_train)

# Transform the training data to include only selected features
X_train_fwd = forward_selector.transform(X_train)

# Transform the test data the same way for consistent evaluation
X_test_fwd = forward_selector.transform(X_test)

# Train the logistic regression model using only the selected features
logreg_fwd = logreg.fit(X_train_fwd, y_train)

# Evaluate the forward-selection model on test data and store results
logreg_fwd_results = evaluate_model("LogReg Forward", logreg_fwd, X_test_fwd, y_test)

# Backward Feature Selection
# --------------------------
# Initialize backward selection strategy
# This starts with all features and removes one feature at a time based on performance
backward_selector = SequentialFeatureSelector(logreg, direction='backward', cv=5)

# Train the backward feature selector on training data
# This identifies which features to remove based on their (lack of) contribution
backward_selector.fit(X_train, y_train)

# Transform training data to exclude removed features
X_train_bwd = backward_selector.transform(X_train)

# Transform test data the same way
X_test_bwd = backward_selector.transform(X_test)

# Train the logistic regression model on the backward-selected features
logreg_bwd = logreg.fit(X_train_bwd, y_train)

# Evaluate the backward-selection model and store results
logreg_bwd_results = evaluate_model("LogReg Backward", logreg_bwd, X_test_bwd, y_test)

# Compare the models and select the better one
# Check which approach (forward or backward) produced better accuracy
if (logreg_fwd_results['Accuracy'] > logreg_bwd_results['Accuracy']):
    logreg_results = logreg_fwd_results  # If forward is better, use its results
else:
    logreg_results = logreg_bwd_results  # Otherwise use backward selection results

# Print results from all methods for comparison
print("Logistic Regression (Forward):", logreg_fwd_results)  # Show forward selection results
print("Logistic Regression (Backward):", logreg_bwd_results)  # Show backward selection results
print("Logistic Regression (Best):", logreg_results)  # Show the better of the two approaches
```

```python
# Import necessary libraries for K-Nearest Neighbors implementation
from sklearn.neighbors import KNeighborsClassifier  # Imports the KNN algorithm
from sklearn.decomposition import PCA  # For dimensionality reduction to visualize high-dimensional data in 2D
from sklearn.model_selection import GridSearchCV  # For hyperparameter tuning
from sklearn.metrics import ConfusionMatrixDisplay  # For visualizing confusion matrix
import numpy as np  # For numerical operations
import matplotlib.pyplot as plt  # For creating visualizations
import seaborn as sns  # For enhanced visualizations
from sklearn.inspection import permutation_importance  # For feature importance analysis
from sklearn.metrics import roc_curve, auc, precision_recall_curve  # For model evaluation metrics
import pandas as pd  # For data manipulation

# Define the parameter grid for hyperparameter tuning
knn_params = {'n_neighbors': [3, 4, 5, 7, 9]}  # Testing different values of k (number of neighbors)
knn = KNeighborsClassifier()  # Initialize the KNN classifier
knn_grid = GridSearchCV(knn, knn_params, cv=5, scoring='accuracy')  # Setup grid search with 5-fold cross-validation
knn_grid.fit(X_train, y_train)  # Train KNN with different parameter combinations on training data

# Get the best model from grid search
best_knn = knn_grid.best_estimator_  # Extract the model with optimal parameters
knn_results = evaluate_model("KNN", best_knn, X_test, y_test)  # Evaluate the best model on test data
print("KNN Best Params:", knn_grid.best_params_)  # Print the optimal parameters
print(knn_results)  # Print evaluation metrics

# Create and display a confusion matrix
plt.figure(figsize=(8, 6))  # Set the figure size
ConfusionMatrixDisplay.from_estimator(knn_grid, X_test, y_test, cmap='Blues')  # Generate confusion matrix
plt.title("Confusion Matrix: KNN")  # Add title
plt.tight_layout()  # Adjust layout
plt.show()  # Display the confusion matrix

# Dimensionality reduction for visualization
pca = PCA(n_components=2)  # Initialize PCA to reduce data to 2 dimensions
X_test_pca = pca.fit_transform(X_test)  # Apply PCA to test data

# Generate predictions and probabilities
knn_preds = best_knn.predict(X_test)  # Get class predictions
knn_prob = best_knn.predict_proba(X_test)[:, 1]  # Get probability of positive class (attrition)

# Visualize predictions vs actual labels using PCA
plt.figure(figsize=(10, 8))  # Set figure size
plt.subplot(2, 1, 1)  # Create first subplot
# Scatter plot of PCA-transformed data colored by predictions
plt.scatter(X_test_pca[:, 0], X_test_pca[:, 1], c=knn_preds, cmap='coolwarm', edgecolor='k', alpha=0.7)
plt.title("KNN Predictions (PCA Projection)")  # Add title
plt.xlabel("Principal Component 1")  # X-axis label
plt.ylabel("Principal Component 2")  # Y-axis label
plt.colorbar(label="Predicted Class")  # Add color legend
plt.grid(True)  # Add grid

plt.subplot(2, 1, 2)  # Create second subplot
# Scatter plot of same PCA-transformed data but colored by actual labels
plt.scatter(X_test_pca[:, 0], X_test_pca[:, 1], c=y_test, cmap='coolwarm', edgecolor='k', alpha=0.7)
plt.title("Actual Labels (PCA Projection)")  # Add title
plt.xlabel("Principal Component 1")  # X-axis label
plt.ylabel("Principal Component 2")  # Y-axis label
plt.colorbar(label="Actual Class")  # Add color legend
plt.grid(True)  # Add grid
plt.tight_layout()  # Adjust layout
plt.show()  # Display the PCA visualization

# Calculate feature importance using permutation importance
# (Since KNN doesn't have built-in feature importance)
result = permutation_importance(
    best_knn, X_test, y_test, n_repeats=10, random_state=42, n_jobs=-1
)  # Calculate importance by shuffling features

# Get feature names (assuming X_test is a DataFrame with column names)
if hasattr(X_test, 'columns'):  # Check if X_test has column names
    feature_names = X_test.columns  # Use column names from DataFrame
else:
    feature_names = [f"Feature {i}" for i in range(X_test.shape[1])]  # Create generic feature names

# Organize feature importance results
perm_importance = pd.DataFrame({
    'Feature': feature_names,
    'Importance': result.importances_mean
})  # Create DataFrame with feature names and importance scores
perm_importance = perm_importance.sort_values('Importance', ascending=False)  # Sort by importance

# Plot top 10 most important features
plt.figure(figsize=(10, 6))  # Set figure size
sns.barplot(x='Importance', y='Feature', data=perm_importance.head(10))  # Create bar plot of top 10 features
plt.title('Top 10 Features for KNN Classifier (Permutation Importance)')  # Add title
plt.xlabel('Mean Decrease in Accuracy')  # X-axis label
plt.tight_layout()  # Adjust layout
plt.show()  # Display feature importance plot

# Generate and plot ROC curve
plt.figure(figsize=(8, 6))  # Set figure size
fpr, tpr, _ = roc_curve(y_test, knn_prob)  # Calculate false positive rate and true positive rate
roc_auc = auc(fpr, tpr)  # Calculate area under ROC curve
# Plot the ROC curve
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
# Plot the diagonal reference line (random classifier)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])  # Set x-axis limits
plt.ylim([0.0, 1.05])  # Set y-axis limits
plt.xlabel('False Positive Rate')  # X-axis label
plt.ylabel('True Positive Rate')  # Y-axis label
plt.title('Receiver Operating Characteristic (ROC) Curve for KNN')  # Add title
plt.legend(loc="lower right")  # Add legend
plt.grid(True)  # Add grid
plt.show()  # Display the ROC curve

# Generate and plot Precision-Recall curve
plt.figure(figsize=(8, 6))  # Set figure size
precision, recall, _ = precision_recall_curve(y_test, knn_prob)  # Calculate precision and recall
plt.plot(recall, precision, color='blue', lw=2)  # Plot precision vs recall
plt.xlabel('Recall')  # X-axis label
plt.ylabel('Precision')  # Y-axis label
plt.title('Precision-Recall Curve for KNN')  # Add title
plt.grid(True)  # Add grid
plt.show()  # Display precision-recall curve

# Analyze the effect of k (number of neighbors) on model performance
k_values = list(range(1, 21))  # Test k values from 1 to 20
train_scores = []  # Initialize list for training scores
test_scores = []  # Initialize list for testing scores

# Loop through each k value
for k in k_values:
    knn_model = KNeighborsClassifier(n_neighbors=k)  # Create KNN model with current k
    knn_model.fit(X_train, y_train)  # Train the model
    train_scores.append(knn_model.score(X_train, y_train))  # Record training accuracy
    test_scores.append(knn_model.score(X_test, y_test))  # Record testing accuracy

# Plot training and testing accuracy vs k
plt.figure(figsize=(10, 6))  # Set figure size
plt.plot(k_values, train_scores, 'o-', color='blue', label='Training Accuracy')  # Plot training accuracy
plt.plot(k_values, test_scores, 'o-', color='orange', label='Testing Accuracy')  # Plot testing accuracy
plt.xlabel('Number of Neighbors (k)')  # X-axis label
plt.ylabel('Accuracy')  # Y-axis label
plt.title('Effect of k on KNN Performance')  # Add title
plt.xticks(k_values)  # Set x-axis ticks
plt.legend()  # Add legend
plt.grid(True)  # Add grid
plt.show()  # Display the plot
```
